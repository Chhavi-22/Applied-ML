```python
import numpy as np
import pandas as pd
```

```python
headers = ["symboling","normalized-losses","make","fuel-type","aspiration", "num-of-doors","body-style",
         "drive-wheels","engine-location","wheel-base", "length","width","height","curb-weight","engine-type",
         "num-of-cylinders", "engine-size","fuel-system","bore","stroke","compression-ratio","horsepower",
         "peak-rpm","city-mpg","highway-mpg","price"]
```

```python
df = pd.read_csv('car_dataset.data', names = headers)
```

```python
df.head()
```

```python
df.replace("?", np.nan, inplace=True)
```

```python
missing = df.isnull()
```

```python
for column in headers:
    print(missing[column].value_counts())
    print(" ")
```

```python
normalize_avg = df["normalized-losses"].astype('float').mean()
normalize_avg
```

```python
df['normalized-losses'].replace(np.nan, normalize_avg, inplace=True)
df['normalized-losses']
```

```python
bore_avg = df['bore'].astype('float').mean()
```

```python
df['bore'].replace(np.nan,bore_avg,inplace = True)
df['bore']
```

```python
stroke_avg = df['stroke'].astype('float').mean()
stroke_avg
```

```python
df['stroke'].replace(np.nan, stroke_avg, inplace = True)
df['stroke'].info()
```

```python
hp_avg = df['horsepower'].astype('float').mean()
hp_avg
```

```python
df['horsepower'].replace(np.nan, hp_avg, inplace = True)
df['horsepower']
```

```python
rpm_avg = df['peak-rpm'].astype("float").mean()
rpm_avg
```

```python
df['peak-rpm'].replace(np.nan, rpm_avg, inplace = True)
df['peak-rpm']
```

```python
df['num-of-doors'].value_counts()
```

```python
df['num-of-doors'].value_counts().idxmax()
```

```python
df['num-of-doors'].replace(np.nan, df['num-of-doors'].value_counts().idxmax(), inplace=True)
```

```python
df.dropna(subset = ['price'], axis = 0, inplace= True)
df.reset_index(drop=True, inplace= True)
```

```python
missing_data = df.isnull()
for column in headers:
    print(missing_data[column].value_counts())
    print(" ")
```
